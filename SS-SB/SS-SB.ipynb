```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.graph_objects as go
```

```python
SS_data = pd.read_csv('E:\SignalModel\SS-SB\SS_data.csv')
SB_data = pd.read_csv('E:\SignalModel\SS-SB\SB_data.csv')
```

```python
# Calculating mean and standard deviation
mean_sigai = SS_data['sigai_output'].mean()
std_sigai = SS_data['sigai_output'].std()
```

```python
# Adjust the definition for the 'signal' levels
# Adjust the definition for the 'signal' levels
SS_data['signal'] = np.where(abs(SS_data['sigai_output'] - mean_sigai) > 1.5 * std_sigai, 'SSS',
                          np.where(abs(SS_data['sigai_output'] - mean_sigai) > 0.7 * std_sigai, 'MSS',
                                   'WSS'))

SS_data.to_csv('E:\SignalModel\SS-SB\Modified_SS_data.csv', index=False)
```

```python
# Updating the scatter plot with corrected signal definitions
fig = go.Figure()

# Adjust color mapping and plotting logic based on the updated criteria
color_map = {'WSB': 'lightgreen', 'MSB': 'green', 'SSB': 'darkgreen'}
for label, color in color_map.items():
    filtered_data = SB_data[SB_data['signal'] == label]
    fig.add_trace(go.Scatter(x=filtered_data['datetime'], y=filtered_data['sigai_output'],
                             mode='markers', name=label, marker=dict(color=color)))

fig.update_layout(
    title='Scatter Plot of sigai_output over Time with Signal Levels',
    xaxis_title='Datetime',
    yaxis_title='sigai_output',
    plot_bgcolor='white'
)

fig.show()
fig.write_image("E:\SignalModel\SS-SB\\SB_Scatter_Plot_with_(0.7, 1.5, 3)_multipliers.png")

```
