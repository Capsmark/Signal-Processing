```python
import pandas as pd
import plotly.graph_objects as go
import warnings

# Suppress future warnings
warnings.simplefilter(action='ignore', category=FutureWarning)

# The directory where your CSV files are stored
data_directory = 'E:\SignalModel\sigai_outputs_h.csv'

# Load the CSV data into a pandas DataFrame
df = pd.read_csv(data_directory, parse_dates=['datetime'] )
df['datetime'] =pd.to_datetime(df['datetime'], yearfirst=True)
df.set_index('datetime', inplace=True)
df.sort_index(ascending=True, inplace=True)
df.describe()
```

```python
# Calculate 24-hour rolling mean
df['rolling_mean'] = df['sigai_output'].rolling('24H').mean()
df
```

```python
# Calculate 24-hour standard deviations
df['std'] = df['sigai_output'].rolling('24H').std()
df
```

```python
df['0.7std'] = 0.7*df['std']
df['1.5std'] = 1.5*df['std']
df['3std'] = 3*df['std']

# Define bounds for the shaded areas
df['upper_0.7std'] = df['rolling_mean'] + df['0.7std']
df['lower_0.7std'] = df['rolling_mean'] - df['0.7std']
df['upper_1.5std'] = df['rolling_mean'] + df['1.5std']
df['lower_1.5std'] = df['rolling_mean'] - df['1.5std']
df['upper_3std'] = df['rolling_mean'] + df['3std']
df['lower_3std'] = df['rolling_mean'] - df['3std']
```

```python
def check_bands(row):
    # Create a list to hold the binary flags for the bands
    bands = [0] * 7
    
    # Check the conditions and set the appropriate band to 1
    if row['sigai_output'] <= row['lower_1.5std']:
        bands[0] = 1
    elif row['lower_1.5std'] <= row['sigai_output'] < row['lower_0.7std']:
        bands[1] = 1
    elif row['lower_0.7std'] <= row['sigai_output'] < row['rolling_mean']:
        bands[2] = 1
    elif row['rolling_mean'] == row['sigai_output'] :  # Assumes a small tolerance for equality
        bands[3] = 1
    elif  row['rolling_mean'] < row['sigai_output'] <= row['upper_0.7std']:
        bands[4] = 1
    elif row['upper_0.7std'] < row['sigai_output'] <= row['upper_1.5std']:
        bands[5] = 1
    elif row['sigai_output'] > row['upper_1.5std']:
        bands[6] = 1
        
    return bands

# Apply the function across each row
new_cols = df.apply(check_bands, axis=1, result_type='expand')
new_cols.columns = ['SB', 'MB', 'WB', 'A', 'WS', 'MS', 'SS']

# Drop any existing columns with these names before the concat operation
df = df.drop(columns=new_cols.columns, errors='ignore')

# Concatenate the new columns with the original DataFrame
df = pd.concat([df, new_cols], axis=1)
df
```

```python
# Plotting
fig = go.Figure()

# Signal Output and Rolling Mean
fig.add_trace(go.Scatter(x=df.index, y=df['sigai_output'], mode='lines', name='Signal Output'))
fig.add_trace(go.Scatter(x=df.index, y=df['rolling_mean'], mode='lines', name='24h Rolling Mean'))

# Standard Deviation Areas
fig.add_trace(go.Scatter(x=df.index, y=df['upper_0.7std'], mode='lines', name='+0.7 STD', line=dict(width=0)))
fig.add_trace(go.Scatter(x=df.index, y=df['lower_0.7std'], mode='lines', name='-0.7 STD', fill='tonexty', line=dict(width=0)))

fig.add_trace(go.Scatter(x=df.index, y=df['upper_1.5std'], mode='lines', name='+1.5 STD', line=dict(width=0)))
fig.add_trace(go.Scatter(x=df.index, y=df['lower_1.5std'], mode='lines', name='-1.5 STD', fill='tonexty', line=dict(width=0)))

fig.add_trace(go.Scatter(x=df.index, y=df['upper_3std'], mode='lines', name='+3 STD', line=dict(width=0)))
fig.add_trace(go.Scatter(x=df.index, y=df['lower_3std'], mode='lines', name='-3 STD', fill='tonexty', line=dict(width=0)))

# Update line colors by accessing the traces and updating the 'line.color' property
for trace in fig.data:
    if trace.name == "-3 STD" or trace.name == "+3 STD":
        trace.line.color = 'rgba(255,0,0,0.5)'  # Red with some transparency for 3 STD lines
    elif trace.name == "-1.5 STD" or trace.name == "+1.5 STD":
        trace.line.color = 'rgba(0,255,0,0.5)'  # Green with some transparency for 2 STD lines
    elif trace.name == "-0.7 STD" or trace.name == "+0.7 STD":
        trace.line.color = 'rgba(0,0,255,0.5)'  # Blue with some transparency for 1 STD lines
    elif trace.name == "24h Rolling Mean":
        trace.line.color = 'red'  # Solid black for the rolling mean
    elif trace.name == "Signal Output":
        trace.line.color = 'gold'  # Solid gold for the signal output

# Update the fill color by setting 'fillcolor' in the case of filled areas
for trace in fig.data:
    if 'fill' in trace:
        trace.fillcolor = 'rgba(200,200,200,0.3)'  # Light grey fill with some transparency

fig.update_layout(title='Signal Output and Rolling Mean with Standard Deviation Regions', xaxis_title='Date', yaxis_title='Value')

fig.show()
```

```python
# Group the data by year and month
grouped = df.groupby([df.index.year, df.index.month])

# Save each group to a separate CSV file
file_paths = []  # List to keep track of the saved file paths
for (year, month), group in grouped:
    file_name = f'E:\SignalModel\monthly_dataset_{year}_{month:02d}.csv'
    group.to_csv(file_name)
    file_paths.append(file_name)
```

```python
# Interpolating to minute-level data
minute_data_frames = []
for file_path in file_paths:
    monthly_data = pd.read_csv(file_path)
    monthly_data['datetime'] = pd.to_datetime(monthly_data['datetime'])
    monthly_data.set_index('datetime', inplace=True)
    monthly_data = monthly_data.resample('T').interpolate()  # 'T' is the code for minute
    minute_data_frames.append(monthly_data)

# Save interpolated data
for i, minute_data in enumerate(minute_data_frames, start=1):
    file_name = f'E:\SignalModel\minute_level_dataset_{i}.csv'
    minute_data.to_csv(file_name)    
```

```python
# Function to save the filtered data for a specific column
def save_filtered_data(df, column_name):
    # If 'datetime' is the index, reset the index first
    if 'datetime' not in df.columns:
        filtered_df = df[df[column_name] == 1].reset_index()[['datetime', 'sigai_output']]
    else:
        # Otherwise, filter as usual
        filtered_df = df[df[column_name] == 1][['datetime', 'sigai_output']]
    
    # Generate the CSV filename based on column name
    filename = f'E:\\SignalModel\\{column_name}_data.csv'
    
    # Save the filtered dataframe as a CSV
    filtered_df.to_csv(filename, index=False)
    
    return filename

# Assuming df is your DataFrame and is already defined
# Replace 'df' with the variable that holds your actual DataFrame.

# List to keep track of generated files
generated_files = {}

# Loop through the columns to be filtered and save the data
for col in ['SS', 'MS', 'WS', 'WB', 'MB', 'SB', 'A']:
    generated_files[col] = save_filtered_data(df, col)

generated_files
```
