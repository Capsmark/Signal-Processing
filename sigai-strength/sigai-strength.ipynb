```python
import pandas as pd
import os
# Base directory for data files and output
base_dir = 'E:\SignalModel'
data_files = [
    'MB_data.csv', 'MS_data.csv', 'SB_data.csv', 
    'SS_data.csv', 'WB_data.csv', 'WS_data.csv'
]

for file_name in data_files:
    # Construct the file path
    file_path = os.path.join(base_dir, file_name)
    df = pd.read_csv(file_path, parse_dates=['datetime'])

    # Calculate min and max of 'sigai_output'
    min_value = df['sigai_output'].min()
    max_value = df['sigai_output'].max()

    # Define the range thresholds
    range_third = (max_value - min_value) / 3
    lower_third_max = min_value + range_third
    middle_third_max = min_value + 2 * range_third

    # Split data into three parts based on 'sigai_output' range
    lower_third_df = df[df['sigai_output'] <= lower_third_max]
    middle_third_df = df[(df['sigai_output'] > lower_third_max) & (df['sigai_output'] <= middle_third_max)]
    upper_third_df = df[df['sigai_output'] > middle_third_max]

    # Directory to save split data files
    output_dir = os.path.join(base_dir, 'sigai-strength', file_name.split('_')[0])
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)

    # Save the split data to separate CSV files
    lower_third_df.to_csv(os.path.join(output_dir, 'lower_third_data.csv'), index=False)
    middle_third_df.to_csv(os.path.join(output_dir, 'middle_third_data.csv'), index=False)
    upper_third_df.to_csv(os.path.join(output_dir, 'upper_third_data.csv'), index=False)

    # Print file paths for confirmation
    print(f'Lower third data saved to {os.path.join(output_dir, "lower_third_data.csv")}')
    print(f'Middle third data saved to {os.path.join(output_dir, "middle_third_data.csv")}')
    print(f'Upper third data saved to {os.path.join(output_dir, "upper_third_data.csv")}')
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import os
# Directory containing the CSV files
directory = 'E:\SignalModel\sigai-strength\WS'
files = [f for f in os.listdir(directory) if f.endswith('.csv')]

# Initialize a DataFrame to store aggregate statistics
aggregate_stats = pd.DataFrame()

for file in files:
    df = pd.read_csv(os.path.join(directory, file))
    
    # Compute descriptive statistics
    stats = df['sigai_output'].describe()
    aggregate_stats[file] = stats
    
    # Plot histogram for 'sigai_output'
    plt.figure(figsize=(10, 6))
    plt.hist(df['sigai_output'], bins=30, alpha=0.7, label=f'{file[:-4]}')  # file[:-4] removes '.csv'
    plt.title(f'Histogram of sigai_output - {file}')
    plt.xlabel('sigai_output')
    plt.ylabel('Frequency')
    plt.grid(True)
    plt.legend()
    plt.close()

# Save the aggregate statistics to a CSV file
aggregate_stats.to_csv('E:\SignalModel\sigai-strength\\aggregate_statistics_WS.csv')

# Print out the path to the summary file
print('Descriptive statistics saved to E:\SignalModel\sigai-strength\\aggregate_statistics.csv')
```
