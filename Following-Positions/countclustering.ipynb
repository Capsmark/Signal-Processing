```python
import pandas as pd

# Load the dataset
file_path = 'E:\SignalModel\Following-Positions\\updated_final_clustered_processed_sigai_buckets.csv'
data = pd.read_csv(file_path)

# Print the column names to confirm the structure
print("Columns in the dataset:", data.columns.tolist())

# Define the columns to retain in each subset
required_columns = ['datetime', 'average', 'trend', 'duration', 'next-signal', 'variance']

# Check if all required columns are present
missing_columns = [col for col in required_columns if col not in data.columns]
if missing_columns:
    raise ValueError(f"Missing columns in the dataset: {missing_columns}")

# Durations from 1 to 18 and Position identifiers
durations = range(1, 19)
positions = ['SS', 'MS', 'WS', 'WB', 'MB', 'SB']

# Dictionary to hold the split datasets
split_datasets = {}

# Split the data based on duration and position
for duration in durations:
    for position in positions:
        # Filter data for the current combination
        filter_mask = (data['duration'] == duration) & (data['signal'] == position)
        filtered_data = data.loc[filter_mask, required_columns]  # Ensure to include required columns

        # Store in dictionary with a key as (duration, position)
        split_datasets[(duration, position)] = filtered_data

        # Optionally, save each filtered dataset to a CSV file
        filtered_data.to_csv(f'E:\SignalModel\Following-Positions\clustering-based-on-count_duration_{duration}_signal_{position}.csv', index=False)

# Example: Access and print the dataset for duration 1 and position SS
if (1, 'SS') in split_datasets:
    print("Example dataset for Duration 1, Position SS:")
    print(split_datasets[(1, 'SS')].head())
else:
    print("No data found for Duration 1, Position SS.")

```
