```python
import pandas as pd
# The directory where your CSV files are stored
data_directory = 'E:\SignalModel\Out_7.csv'

# Load the CSV data into a pandas DataFrame
df = pd.read_csv(data_directory, parse_dates=['datetime'] )
```

```python
# Print DataFrame columns to check what's available
print("Columns in DataFrame:", df.columns.tolist())

# Define the pairs of columns to check
pairs = ['SB', 'MB', 'WB', 'WS', 'MS', 'SS']

output_files = []  # To track output files

# Check if all required columns are in the DataFrame
required_columns = set(pairs + ['datetime', 'sigai_output'])
missing_columns = required_columns - set(df.columns)
if missing_columns:
    raise ValueError(f"DataFrame is missing the following required columns: {missing_columns}")

for first in pairs:
    for second in pairs:
        # Collect qualifying row indices
        indices = [i for i in range(len(df) - 1) if df.loc[i, first] == 1 and df.loc[i + 1, second] == 1]
        
        # Create a subset DataFrame if indices are found
        if indices:
            subset_df = df.loc[indices + [i + 1 for i in indices], ['datetime', 'sigai_output']].reset_index(drop=True)
            
            # Save to CSV if the subset DataFrame is not empty
            if not subset_df.empty:
                output_filename = f'\\SignalModel\\{first}_to_{second}_output.csv'
                subset_df.to_csv(output_filename, index=False)
                output_files.append(output_filename)

output_files  # List of created CSV files

```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt

# Directory containing the CSV files
directory = 'E:\SignalModel\segmentation'
files = [f for f in os.listdir(directory) if f.endswith('.csv')]

# Initialize a DataFrame to store aggregate statistics
aggregate_stats = pd.DataFrame()

for file in files:
    df = pd.read_csv(os.path.join(directory, file))
    
    # Compute descriptive statistics
    stats = df['sigai_output'].describe()
    aggregate_stats[file] = stats
    
    # Plot histogram for 'sigai_output'
    plt.figure(figsize=(10, 6))
    plt.hist(df['sigai_output'], bins=30, alpha=0.7, label=f'{file[:-4]}')  # file[:-4] removes '.csv'
    plt.title(f'Histogram of sigai_output - {file}')
    plt.xlabel('sigai_output')
    plt.ylabel('Frequency')
    plt.grid(True)
    plt.legend()
    plt.close()

# Save the aggregate statistics to a CSV file
aggregate_stats.to_csv('E:\\SignalModel\\segmentation\\aggregate_statistics.csv')

# Print out the path to the summary file
print('Descriptive statistics saved to E:\SignalModel\segmentation\\aggregate_statistics.csv')

```
