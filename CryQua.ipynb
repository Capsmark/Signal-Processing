```python
import requests
import pandas as pd
from datetime import datetime, timedelta
import io

API_KEY = 'EUfMfAM6TILIl5UUvsONlrG2DPYfrQsdWVnNv2gw'
ENDPOINT = "https://api.cryptoquant.com/v1/btc/market-data/price-ohlcv"

def get_data(start_date, end_date):
    params = {
        'symbol': 'btc_usdt',
        'exchange': 'binance',
        'window': 'min',
        'market': 'perpetual',
        'format': 'csv',
        'from': start_date.strftime('%Y%m%dT%H%M%S'),
        'to': end_date.strftime('%Y%m%dT%H%M%S'),
        'limit': '100000',
        'api_key': API_KEY
    }

    response = requests.get(ENDPOINT, params=params)
    if response.status_code == 200:
        return pd.read_csv(io.StringIO(response.content.decode('utf-8')))
    else:
        print("Error:", response.status_code)
        return pd.DataFrame()

def merge_data(start_date, end_date, interval_days=15):
    all_data = pd.DataFrame()
    current_start_date = start_date

    while current_start_date < end_date:
        current_end_date = min(current_start_date + timedelta(days=interval_days), end_date)
        data = get_data(current_start_date, current_end_date)
        all_data = pd.concat([all_data, data]).drop_duplicates().reset_index(drop=True)
        current_start_date = current_end_date
        
    all_data = all_data.sort_values(by='datetime')  # Adjust this to the correct column name
    return all_data

# Define date range
start_date = datetime(2023, 6, 1)
end_date = datetime(2024, 8, 4)

# Merge data for the given range
merged_data = merge_data(start_date, end_date)

# Save the merged data to CSV
merged_data.to_csv('price_2023-06-01_to_2024-08-4_min.csv', index=False)

print("Price data saved to price_2023-06-01_to_2024-08-04_min.csv")
```
