```python
import pandas as pd
import plotly.graph_objs as go
from plotly.subplots import make_subplots

# Load the dataset from the provided Excel file
data = pd.read_csv('E:\SignalModel\Out_7.csv')
# Convert 'datetime' to datetime objects
data['datetime'] =pd.to_datetime(data['datetime'], yearfirst=True)
data.set_index('datetime', inplace=True)
data.sort_index(ascending=True, inplace=True)
print("data loaded")
```

```python
# Calculate 24 hour rolling average of sigai_output
data['24h_rolling_mean'] = data['sigai_output'].rolling(window=24).mean()
print("moving average calculated")
```

```python
# Calculate 24 hour rolling std of sigai_output
data['24h_rolling_std'] = data['sigai_output'].rolling(window=24).std()
print("moving standard deviation calculated")
```

```python
# Calculate upper and lower bands
std_multipliers = [0.7, 1.5, 3]
for multiplier in std_multipliers:
    data[f'{multiplier} STD Upper'] = data['24h_rolling_mean'] + (data['24h_rolling_std'] * multiplier)
    data[f'{multiplier} STD Lower'] = data['24h_rolling_mean'] - (data['24h_rolling_std'] * multiplier)
    print("std ranges calculated")
```

```python
# Define RGBA colors for each column
colors = {
    'SB': 'rgba(0, 153, 0, 0.2)',   # Dark Green
    'MB': 'rgba(0, 204, 0, 0.2)',   # Medium Green
    'WB': 'rgba(0, 255, 0, 0.2)',   # Light Green
    'WS': 'rgba(255, 204, 204, 0.2)', # Light Red
    'MS': 'rgba(255, 102, 102, 0.2)', # Medium Red
    'SS': 'rgba(255, 0, 0, 0.2)'}     # Dark Red

std_band_colors = {
    3: 'rgba(150, 150, 150, 1)',     # Even darker grey for 3 STD
    1.5: 'rgba(60, 60, 60, 1)',   # Darker grey for 1.5 STD
    0.7: 'rgba(10, 10, 10, 1)'    # Dark grey for 0.7 STD
}
y_margin = 0.0005
x_margin = pd.Timedelta(minutes=30)
print("colors and margins have been updated")
```

```python
# Create a figure with secondary y-axis
fig = make_subplots(specs=[[{"secondary_y": True}]])

# Plot the mean and std lines
fig.add_trace(go.Scatter(
    x=data.index,
    y=data['24h_rolling_mean'],
    name='24h Rolling Mean',
    line=dict(color='rgba(255,255,255,0.4)', dash="dot")
))

for multiplier in [3, 1.5, 0.7]:
    fig.add_trace(go.Scatter(
        x=data.index, y=data[f'{multiplier} STD Upper'],
        line=dict(width=2, color=f"rgba(255,0,0,{0.3 * multiplier})"),
        name=f'+{multiplier} STD',
        showlegend=True,
    ))
    fig.add_trace(go.Scatter(
        x=data.index, y=data[f'{multiplier} STD Lower'],
        line=dict(width=2, color=f"rgba(0,255,0,{0.3 * multiplier})"),
        name=f'-{multiplier} STD',
        showlegend=True
    ))
    print("mean and std have been plotted")
```

```python
# Create scatter traces for dots and shapes for boxes
for col, color in colors.items():
    # Add scatter trace for dots
    fig.add_trace(go.Scatter(
        x=data[data[col] == 1].index, 
        y=data[data[col] == 1]['sigai_output'],
        mode='markers',
        marker=dict(color=color),
        name=col
    ))

print("scatter traces have been plotted")    
```

```python
# Draw boxes around consecutive ones with matching color
for col, color in colors.items():
    current_patch_start = None
    for idx, val in enumerate(data[col]):
        if val == 1 and current_patch_start is None:
            current_patch_start = idx
        elif val == 0 and current_patch_start is not None:
            if idx - current_patch_start > 1:
                y_values = data.iloc[current_patch_start:idx]['sigai_output']
                y0, y1 = y_values.min() - y_margin, y_values.max() + y_margin  # Apply margin
                x0 = data.index[current_patch_start] - x_margin  # Apply x-margin
                x1 = data.index[idx - 1] + x_margin  # Apply x-margin
                fig.add_shape(
                    type="rect",
                    x0=x0, y0=y0,
                    x1=x1, y1=y1,
                    line=dict(color=color),
                    fillcolor=colors[col],  # Use the RGBA colors
                    layer="below"
                )
            current_patch_start = None
            
print("boxes have been plotted")            
```

```python
# Set figure layout
fig.update_layout(title="Scatter Plot with Color-coded Dots and Boxes",
                  xaxis_title="Datetime",
                  yaxis_title="Sigai Output",  width=1000, height=600)

# Show figure
fig.show()
```
